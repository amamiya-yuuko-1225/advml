### The GridWorld Environment

We already provide a basic implementation of the grid world environment which you will use,
as well as the constants `ACTION_LEFT, ACTION_RIGHT, ACTION_UP, ACTION_DOWN`.

```python
#DO NOT MOIFY
ACTION_LEFT = 0
ACTION_RIGHT = 1
ACTION_UP = 2
ACTION_DOWN = 3

def clip(x, a, b):
    if x < a:
        return a
    if x > b:
        return b
    
    return x

class GridWorld:
    
    def __init__(self, size=3):
        self.size = size
        self.state = [0,0]
        self.actions = [ACTION_LEFT, ACTION_RIGHT, ACTION_UP, ACTION_DOWN]
        
    def reset_world(self):
        self.state = [0,0]
    
    def is_terminal(self):
        return abs(self.state[0]) == self.size and abs(self.state[1]) == self.size
    
    def get_current_reward(self):
        if self.state == [self.size, self.size] or \
           self.state == [-self.size, -self.size]:
            return -1
        elif self.state == [-self.size, self.size] or \
             self.state == [self.size, -self.size]:
            return +1
        else:
            return 0
        
    def list_actions(self):
        return self.actions
    
    def step(self, action):
        if action not in self.actions:
            raise NotImplementedError("Action not defined")
            
        if self.is_terminal():
            raise NotImplementedError("Epsiode terminated! Call reset_world first.")

        if action == ACTION_LEFT:
            self.state[0] = clip(self.state[0]-1, -self.size, self.size)
            
        elif action == ACTION_RIGHT:
            self.state[0] = clip(self.state[0]+1, -self.size, self.size)
            
        elif action == ACTION_UP:
            self.state[1] = clip(self.state[1]+1, -self.size, self.size)
            
        elif action == ACTION_DOWN:
            self.state[1] = clip(self.state[1]-1, -self.size, self.size)
        
    def get_coordinates(self):
        s = np.array(self.state)
        s[0] = s[0] + self.size
        s[1] = self.size - s[1]
        
        return s
    
    def get_state(self):
        return self.get_coordinates()[0] + \
               self.get_coordinates()[1] * (self.size * 2 + 1)
    
    def render_state(self):
        N = self.size
        x = self.state[0]
        y = self.state[1]

        fig, ax = plt.subplots(figsize=(5,5))
        ax.set_axis_off()

        tb = Table(ax, bbox=[0,0, 1, 1])

        row_labels = np.arange(2 * N + 1)
        col_labels = np.arange(2 * N + 1)
        width, height = .5, .5

        values = np.zeros((N*2+1, N*2+1))
        values[0,0] = values[-1, -1] = 1
        values[-1,0] = values[0, -1] = -1

        cell_colors = ('white', 'whitesmoke')
        font_colors = ('black', 'white')

        for (row_idx, col_idx), cell_val in np.ndenumerate(values):
            idx = (col_idx + row_idx) % 2
            if cell_val !=  0.:
                text = "%.1f" % cell_val
                if cell_val > 0:
                    c = "lightblue"
                else:
                    c = "lightcoral"
            elif row_idx == N and col_idx == N:
                text = "start"
                c = "antiquewhite"
            else:
                text = ""
                c = cell_colors[idx]

            tb.add_cell(row_idx, col_idx, width, height,
                        text=text,
                        loc='center',
                        facecolor=c)

        for (row_idx, col_idx), cell_val in np.ndenumerate(values):
            idx = (col_idx + row_idx) % 2
            tb._cells[(row_idx, col_idx)]._text.set_color(font_colors[idx])

        pos = plt.Circle((0.5 + 1/(2 * N + 1) * x, 0.5 + 1/(2 * N + 1) * y + .04), 0.01, color='black')
        ax.add_patch(pos)
        ax.add_table(tb)
```

In this part, we will define the policy according to an exponential softmax distribution where action preferences are linear using feature vectors $\phi_{s,a}$:
$$
\pi_\theta(a|s) = \frac{\exp{\theta^\top \phi_{s,a}}}{\sum_{a'} {\exp{\theta^\top \phi_{s,a'}}}}
$$

**Hint:** The function `direct_parameterisation_derivative` should return a vector of size $4 |S| = 4 \times 7^2=196$, since you have 4 actions and $7^2$ states.

**Hint:** Only 4 entries of $\frac{\partial \pi_\theta(a|s)}{\partial \theta} \in \mathbb{R}^{196}$ should be nonzero for fixed $a,s$.

$$\pi_\theta(a|s) = \frac{\exp(\theta_{s,a})}{\sum_{a'} \exp(\theta_{s,a'})}$$
where $\theta_{s,a}$ is the entry of $\theta$ corresponding to state $s$ and action $a$.



$$\frac{\partial \pi_\theta(a|s)}{\partial \theta_{s',a'}} = 

\begin{cases}
\pi_\theta(a|s) \cdot (1 - \pi_\theta(a|s)) & \text{if } s'=s, a'=a \\
-\pi_\theta(a|s) \cdot \pi_\theta(a'|s) & \text{if } s'=s, a' \neq a \\
0 & \text{if } s' \neq s
\end{cases}$$


```python
def direct_parameterisation_gradient(action: int, state: int, theta: np.ndarray):
    """Returns the partial derivatives of the policy with respect to theta
    
    Assume actions are numbered from 0 to 3, and states are numbered from 0 to 48
    as explained above.
    """

    # TODO: FILL IN YOUR CODE HERE!


    gradient = np.zeros(196)
    
    # we need \pi(a^\prime|s) for all four actions a^\prime given state s
    preferences = np.zeros(4)
    for a in range(4):
        phi = get_features_onehot(state, a)
        preferences[a] = np.dot(theta, phi)
    
    # calculate softmax probabilities by normalizing exponentiated preferences
    exp_preferences = np.exp(preferences)
    action_probs = exp_preferences / np.sum(exp_preferences)
    
    
    # \pi( a | s )
    pi_a = action_probs[action]
    
    # only 4 entries of the gradient are non-zero
    for a in range(4):
        index = 4 * state + a
        if a == action:
            # \partial pi(a|s) / \partial theta
            gradient[index] = pi_a * (1 - pi_a)
        else:
            # \partial pi(a^\prime|s) / \partial theta, a^\prime != a
            gradient[index] = -pi_a * action_probs[a]
    
    return gradient
    
```

### REINFORCE Implementation

**Task 1.4 (1 pt):** In the lectures, you have been introduced to the REINFORCE algorithm. Explain in a few sentences what the regular REINFORCE objective aims to optimise and how it is achieved only using sample and no prior information on the MDP dynamics. 

*Write in the Markdown cell below (use LaTeX-math mode for equations, etc.).*

**Answear:**
The algorithm aims to optimize $\pi(a|s, \theta)$, which is the policy function parameterized by $\theta$. The objective is to maximize the expected return. The optimization is achieved using the policy gradient theorem, which states that the gradient of the expected return can be expressed as $E_{\pi}[\nabla_\theta \log \pi(a|s, \theta) G_t]$. This allows us to update the policy parameters $\theta$ in the direction of the gradient using sampled trajectories from the environment, without requiring knowledge of the MDP dynamics.

**Task 1.5 (2 pts):** Using your functions implemented in Task 1.1, 1.2 and 1.3, fill in the `ReinforceLearner` class. You can use the already implemented methods of the GridWorld class:

```python
class ReinforceLearner:

    def __init__(self, learning_rate, gamma):
        """Intialisation code"""
        
        self.world = GridWorld()
        self.learning_rate = learning_rate
        self.gamma = gamma
        self.theta = get_initial_parameters()
        
        ## TODO: you may use other variables or inputs as you hyper-parameters
        
    def get_action_prob(self, state:int):
        """Given the state, return the probability of taking each of the 4 actions"""
        
        ## TODO: FILL IN YOUR CODE HERE

        # Just use the formula for softmax policy
        preferences = np.zeros(4)
        for action in range(4):
            phi = get_features_onehot(state, action)
            preferences[action] = np.dot(self.theta, phi)

        preferences = preferences - np.max(preferences)
        exp_preferences = np.exp(preferences)
        action_probs = exp_preferences / np.sum(exp_preferences)
        
        return action_probs

        
    
    def train(self, n_updates:int, n_trajectories:int):
        """Run the training procedure for n_updates steps
        
        n_trajectories: how many trajectories should be sampled each REINFORCE step
        n_updates: how many parameters updates to make (each parameter update would sample n_trajectories)"""
        
        rewards = []
        for episode in range(n_updates):
            rewards.append(self.reinforce_step(n_trajectories))
                
        return rewards
        
    def reinforce_step(self, n_trajectories: int):
        """Run a single gradient step for REINFORCE (without baseline) using n_trajectories many trajectories for estimation.
        
        Also return the average reward observed"""
        
        ## TODO: FILL IN YOUR CODE HERE

        # store cumulative gradient over all sampled trajectories
        gradient_sum = np.zeros(196)
        total_rewards = []
        
        # sample n_trajectories many trajectories
        for traj_idx in range(n_trajectories):
            self.world.reset_world()
            
            # store single trajectory information
            states = []
            actions = []
            rewards = []
            
            # generate a trajectory
            while not self.world.is_terminal():
                
                current_state = self.world.get_state()
                
                # select action according to current policy
                action_probs = self.get_action_prob(current_state)
                action = np.random.choice(4, p=action_probs)
                
                
                states.append(current_state)
                actions.append(action)
                
                # take action and observe reward
                self.world.step(action)
                
                
                reward = self.world.get_current_reward()
                rewards.append(reward)
            
            
            total_rewards.append(sum(rewards))
            
            # compute the gradient contribution from this trajectory
            T = len(states)  
            
            for t in range(T):
                # calculate G_t from time t
                G_t = 0.0
                for k in range(t, T):
                    G_t += (self.gamma ** (k - t)) * rewards[k]
                
                # calculate \nabla log π(a_t|s_t)
                # first compute \nabla π(a_t|s_t)
                # then according to chain rule get \nablalog π(a_t|s_t) by dividing by π(a|s)
                grad_log_pi = direct_parameterisation_gradient(
                    actions[t], states[t], self.theta
                )
                pi_a = self.get_action_prob(states[t])[actions[t]]
                grad_log_pi = grad_log_pi / pi_a
                
                # accumulate gradient contribution
                gradient_sum += grad_log_pi * G_t
        
        # calculate average gradient over all trajectories
        gradient = gradient_sum / n_trajectories
        
        self.theta += self.learning_rate * gradient
    
        return np.mean(total_rewards)
        
        
    def evaluate(self, render: bool = True):
        """Evaluate current algorithm for a single episode and return.
        
        You should 
        (a) Print the total reward from the episode.
        (b) Visualise the whole state history. (Hint: You can use multiple calls to GridWorld.render_state())"""
        
        ## TODO: FILL IN YOUR CODE HERE

        self.world.reset_world()
        
        # store state history for visualization
        states_history = [self.world.state.copy()]  
        total_reward = 0
         
        while not self.world.is_terminal():
            current_state = self.world.get_state()
            
            # greedily select action
            action_probs = self.get_action_prob(current_state)
            action = np.argmax(action_probs)
  
            self.world.step(action)
            states_history.append(self.world.state.copy())
            

            reward = self.world.get_current_reward()
            total_reward += reward
        

        print(f"Total reward: {total_reward}")
        print(f"Trajectory length: {len(states_history)}")
        
        # visualize state history
        if render:
            print("\nState history visualization:")
            for i, state in enumerate(states_history):
                print(f"\nStep {i}:")
                self.world.state = state
                self.world.render_state()
        
        pass
```

### Advantage Actor-Critic (A2C)

A more stable alternative to vanilla policy gradients via REINFORCE is the so-called A2C algorithm, which is a actor-critic method.
First, you should familiarise yourself with the A2C algorithm reading the original [paper](https://arxiv.org/pdf/1602.01783.pdf).

**Task 2.1 (3 pts):** Explain in a few sentences each, 

1. how the A2C algorithm addresses the shortcomings of vanilla policy gradient methods,
2. the difference between A2C and A3C,
3. why the A3C algorithm could be expected to perform even better than A2C.

*Write in the Markdown cell below (use LaTeX-math mode for equations, etc.).*

**Answear:**
1. Vanilla policy gradient methods suffer from high variance because it rely solely on sampled returns $R_t$. With high variance, the learning process will be highly unstable, converge slowly, and be extremely sensitive to hyperparameters such as the learning rate. A2C addresses this issue by introducing a "Critic." The Critic's role is to learn a value function V(s) that estimates the expected return under state $s$ .With this value function, we can calculate the "Advantage Function," defined as $A(s, a) = Q(s, a) - V(s)$. This function measures how much better taking a specific action $a$ is, compared to the baseline (i.e.,$V(s)$)

2. A2C is a synchronous algorithm where multiple agents collect experience in parallel within their respective environment replicas. During an update step, it waits for all agents to complete collecting a specified amount of experience before aggregating these experiences for a single batch gradient update. A3C on the other hand, uses multiple agents running in separate threads, independently interacts with the environment and calculates gradients. Subsequently, they asynchronously apply their computed gradients to a globally shared model parameter. A2C is more GPU-friendly, while A3C is designed for CPU-based parallelism.

3. Since each agents explores different parts of the environment, the experiences they gather are diverse. The asynchronous update mechanism ensures that gradients applied to the global model originate from experiences at different times and states. This "randomize" gradient update process can reduces the correlation between data points. Thus, making the training more stable and avoid overfitting.

### Entropy regularized A2C

The exploration-exploition trade-off is also a significant problem to be solved in the context of policy gradients.
There have been numerous approaches in literature aimed towards maximizing exploration in a principled and theoretically meaningful manner.
One such example is the entropy regularised policy gradient objective:
$$
J_\tau(\theta) = J(\theta) + \tau \mathbb{E}_{s \sim d^\pi_s}\left[-\sum_a \pi_\theta(a|s) \log \pi_\theta(a|s)\right],
$$
where $J(\theta)$ is the vanilla policy gradient objective and $\tau$ is a hyperparameter controlling regularisation.

**Task 2.4 (0.5 pt):** Explain why entropy regularization is relevant for encouraging exploration.

*Write in the Markdown cell below (use LaTeX-math mode for equations, etc.).*

**Answear:**

In reinforcement learning, the goal of policy gradient methods is to maximise the expected return:
$$J(\theta) = \mathbb{E}_{\pi_\theta} \left[ \sum_t r_t \right]$$
However, directly optimising this objective often leads the policy to become overly deterministic—favoring a single action too early. This causes several issues such as insufficient exploration, missing better strategies or premature convergence to local optima. To address this, **entropy regularisation** is introduced. Entropy measures the randomness of a policy:
$$
[-\sum_a \pi_\theta(a|s) \log \pi_\theta(a|s)]
$$
A high-entropy policy is more random and encourages exploration, while a low-entropy policy is more deterministic and favors exploitation.By rewarding the agent for maintaining uncertainty, entropy regularisation helps prevent premature convergence and promotes more effective exploration throughout training.
